### Train model from beginning
Specify which YOLO model (YOLO11medium)

```python
model = YOLO("yolo11m.pt")
```

Path to downloaded training dataset

```python
data_path = os.path.join(dataset.location, "data.yaml")
```

Train model

```python
results = model.train(
    data=data_path,
    device='cuda', epochs=200, imgsz=1280,
    batch=8, cache=True, amp=True, workers=2,
    degrees=180, flipud=0.5, fliplr=0.5, mosaic=1.0,
    patience=50, lr0=0.001
)
```
