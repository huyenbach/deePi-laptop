```python
import subprocess
import os
from pathlib import Path
```

```python
# 1. Setup your paths
base_dir = Path.cwd()
video_folder = base_dir / "simultaneous-detections"
input_file = video_folder / "10.0.12.2_202307070748.h264"
output_file = video_folder / "10.0.12.2_202307070748.mp4"

# 2. Match original camera's frame rate setting (fps 25)
CAMERA_FPS = "6"

# 3. Build the strict translation command
command = [
    'ffmpeg',
    '-r', CAMERA_FPS,          
    '-i', str(input_file),
    '-c:v', 'copy',            
    '-y',                      
    str(output_file)
]
subprocess.run(command,check=True)
```

### Define conversion function

```python
def convert_to_mp4(input_file, output_file):
    CAMERA_FPS = "24"
    
    command = [
        'ffmpeg',
        '-r', CAMERA_FPS,          
        '-i', str(input_file),
        '-c:v', 'copy',            
        '-y',                      
        str(output_file)
    ]
    
    try:
        subprocess.run(command, check=True)
        print(f"Conversion successful: {output_file}")
    except subprocess.CalledProcessError as e:
        print("Error during conversion:",e)
```

### Build path for loops

```python
subdirs = os.listdir('../happel mooring/video data/')
subdirs = [s for s in subdirs if s!= '.DS_Store']
subdirs
```

```python
data_path = '../happel mooring/video data/' + subdirs[2] + '/Videos/'
files = os.listdir(data_path)
print(data_path)
print(len(files))
```

### Build output directories and convert all videos

```python
for s in range(len(subdirs)-1):
    data_path = '../happel mooring/video data/' + subdirs[1] + '/Videos/'
    files = os.listdir(data_path)
    print(data_path)
    print(len(files))

    #os.makedirs('./converted_videos/', exist_ok=True)
    os.makedirs('./converted_videos/' + subdirs[1] + "/", exist_ok=True)
    for i in files:
        inPath = data_path + i
        outPath = './converted_videos/' + subdirs[1] + "/" + i.replace('h264','mp4')
        convert_to_mp4(inPath, outPath)
    
```
