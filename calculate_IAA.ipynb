calculate percent agreement, Cohen's kappa, and confusion matrix

Load annotation csv files 

```python
ann1 = pd.read_csv("/Users/alopias/Desktop/Huyen-deePi/deePi-laptop/active_review_batch_003/current_labels.csv") # Huyen's labels
ann2 = pd.read_csv("brendan_labels_v2.csv")        # Brendan's labels

# Rename label columns
ann1 = ann1.rename(columns={"manual_label": "label_1"})         # Huyen
ann2 = ann2.rename(columns={"manual_label": "label_2"})         # Brendan

# Merge by image name to make sure labels line up correctly
df = ann1.merge(ann2, on="image_name", how="inner")
```

Calculate IAA and make confusion matrix

```python
# P0
percent_agreement = (df["label_1"] == df["label_2"]).mean()
print("Percent agreement:", round(percent_agreement,3))

# Kappa
kappa = cohen_kappa_score(df["label_1"], df["label_2"])
print("Cohen's kappa:", round(kappa,3))

```

Confusion matrix

```python
labels = ["particulate","animal"]
cm = confusion_matrix(df["label_1"], df["label_2"], labels=labels)
cm_df = pd.DataFrame(cm,
                     index=[f"Huyen: {x}" for x in labels],
                     columns=[f"Brendan: {x}" for x in labels])
print(cm_df)
```
